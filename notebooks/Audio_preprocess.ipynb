```python
import os
import subprocess

root = "/content/drive/MyDrive/AnimalSoundDataset"

for root_dir, _, files in os.walk(root):
    for file in files:
        if file.endswith((".mp3", ".aac")):
            input_path = os.path.join(root_dir, file)
            output_path = os.path.splitext(input_path)[0] + ".wav"

            cmd = [
                "ffmpeg", "-i", input_path,
                "-ac", "1",
                "-ar", "16000",
                output_path
            ]
            print(cmd)
            subprocess.run(cmd)
```

```python
import os
import librosa
import soundfile as sf
import numpy as np

TARGET_DURATION = 4.0  # seconds
SAMPLE_RATE = 16000
TARGET_LENGTH = int(TARGET_DURATION * SAMPLE_RATE)

for root, _, files in os.walk(root):
    for file in files:
        if file.lower().endswith(".wav"):
            file_path = os.path.join(root, file)

            # Load audio (force mono + 16kHz)
            audio, sr = librosa.load(file_path, sr=SAMPLE_RATE, mono=True)

            if len(audio) > TARGET_LENGTH:
                # Trim
                audio = audio[:TARGET_LENGTH]
            else:
                # Pad
                padding = TARGET_LENGTH - len(audio)
                audio = np.pad(audio, (0, padding), mode='constant')

            # Save back
            sf.write(file_path, audio, SAMPLE_RATE)

print("All files trimmed/padded to exactly {D} seconds.")
```
